# Draw Operators

```python
import cv_depot.api as cvd
```

<font size=4 color="#EBB483"><br>**swatch** draws a color swatch.

```python
image = cvd.ops.draw.swatch((512, 512, 3), cvd.BasicColor.BLUE1)
image
```

<font size=4 color="#EBB483"><br>**grid** draws a grid over a given image.

```python
cvd.ops.draw.grid(
    image, 
    shape=(4, 4),
    color=cvd.BasicColor.RED2,
    thickness=2,
)
```

<font size=4 color="#EBB483"><br>**checkerboard** draws a checkerboard.

```python
cvd.ops.draw.checkerboard(4, 2, (128, 128))
```

<font size=4 color="#EBB483"><br>**outline** outlines an image given a single mask channel.

```python
src = '/home/ubuntu/images/misc/p-ring_s-rndr001_d-demo_v002_f0300.exr'
image = cvd.Image.read(src)
cvd.ops.draw.outline(image, mask='a', width=4)[:, :, 'rgb']
```

<font size=4 color="#EBB483"><br>**annotate** outlines and colors an image given a single mask channel.

```python
cvd.ops.draw.annotate(image, mask='a', width=4, opacity=0.4)
```
